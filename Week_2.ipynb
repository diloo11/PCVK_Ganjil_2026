```python
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("Wahyu KTM.jpeg")

# sensor pertama
for y in range(300, 850):
    for x in range(140, 200):
        img[y, x] = [255, 255, 255]

# sensor kedua
for y in range(300, 600):
    for x in range(100, 130):
        img[y, x] = [255, 255, 255]

# sensor ketiga
for y in range(300, 750):
    for x in range(60, 95):
        img[y, x] = [255, 255, 255]

cv2_imshow(img)

```
