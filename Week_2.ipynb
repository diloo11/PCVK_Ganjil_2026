```python
import numpy as np
import matplotlib.pyplot as plt

# Membuat kanvas hitam 512 x 512
black_int16 = np.zeros((512, 512, 3), dtype=np.int16)
black_int32 = np.zeros((512, 512, 3), dtype=np.int32)

# Menampilkan dtype, min, max, dan ukuran memori
print("=== INT16 ===")
print("dtype :", black_int16.dtype)
print("min   :", black_int16.min())
print("max   :", black_int16.max())
print("nbytes:", black_int16.nbytes)

print("\n=== INT32 ===")
print("dtype :", black_int32.dtype)
print("min   :", black_int32.min())
print("max   :", black_int32.max())
print("nbytes:", black_int32.nbytes)

# Menampilkan kedua kanvas
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(black_int16)
plt.title("int16")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(black_int32)
plt.title("int32")
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("KarinaCard.jpeg")

# Convert BGR to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Red-Blue: keep R and B, remove G
red_blue = img_rgb.copy()
red_blue[:, :, 1] = 0

# Green-Blue: keep G and B, remove R
green_blue = img_rgb.copy()
green_blue[:, :, 0] = 0

# Display side by side
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(red_blue)
plt.title("Red-Blue Channels")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(green_blue)
plt.title("Green-Blue Channels")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("KarinaCard.jpeg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Vertical flip
flip_vertical = cv2.flip(img_rgb, 0)

# Horizontal flip
flip_horizontal = cv2.flip(img_rgb, 1)

# Both vertical and horizontal
flip_both = cv2.flip(img_rgb, -1)

# Display
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(flip_vertical)
plt.title("Vertical Reflection")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(flip_horizontal)
plt.title("Horizontal Reflection")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(flip_both)
plt.title("Vertical + Horizontal Reflection")
plt.axis("off")

plt.tight_layout()
plt.show()
```

Practicum Task Analysis

Question

“What percentage of pixels in an image have an intensity value below 100?”

Program Design

The program reads the image, converts it to grayscale, counts the pixels with intensity below 100, and then calculates the percentage.

```python
import cv2
import numpy as np

# Read the image
img = cv2.imread("DiloActivity.jpeg")

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Count pixels with intensity below 100
dark_pixels = np.sum(gray < 100)

# Total number of pixels
total_pixels = gray.size

# Calculate percentage
percentage = (dark_pixels / total_pixels) * 100

print("Total pixels       :", total_pixels)
print("Pixels below 100   :", dark_pixels)
print("Percentage         :", percentage, "%")
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read image
img = cv2.imread("Parfume.jpeg")

# Get image dimensions
h, w = img.shape[:2]

# Determine square canvas size
if h > w:
    size = h
else:
    size = w

# Resize image while maintaining original ratio
if h > w:
    new_h = size
    new_w = int(w * size / h)
else:
    new_w = size
    new_h = int(h * size / w)

resized = cv2.resize(img, (new_w, new_h))

# Create white square canvas
canvas = np.zeros((size, size, 3), dtype=np.uint8)
canvas[:, :] = (255, 255, 255)

# Calculate position to center the image
x = int((size - new_w) / 2)
y = int((size - new_h) / 2)

# Put image in the center
canvas[y:y+new_h, x:x+new_w] = resized

# Add store name
font_size = size / 600
thickness = int(size / 300)

cv2.putText(
    canvas,
    "TOKO DILO",
    (int(size * 0.08), int(size * 0.95)),
    cv2.FONT_HERSHEY_SIMPLEX,
    font_size,
    (0, 0, 0),
    thickness
)

# Display result
cv2_imshow(canvas)
```
